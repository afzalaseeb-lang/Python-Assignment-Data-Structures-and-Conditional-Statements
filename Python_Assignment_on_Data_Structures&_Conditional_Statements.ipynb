```python
#Create a list named age_list with five integer elements.
age_list = [24, 25, 26, 27, 28]
print(age_list)
#Create a list named name_list with five string elements.
name_list = ["Aseeb", "Rahul", "Arun", "Vishnu", "Anand"]
print(name_list)
```

```python
#Append the string "Yazhini" to name_list.
name_list = ["Aseeb", "Rahul", "Arun", "Vishnu", "Anand"]
name_list.append("Yazhini")
print(name_list)

#Insert the element 30 at index 2 in age_list.
age_list = [24, 25, 26, 27, 28]
age_list.insert(2, 30)
print(age_list)

#Remove the string "Yazhini" from name_list.
name_list.remove("Yazhini")
print(name_list)

#Pop the last element from age_list.
age_list.pop()
print(age_list)

#Extend the age_list with additional ages [29, 30, 26].
age_list.extend([29, 30, 26])
print(age_list)

#Sort age_list in descending order.
age_list.sort(reverse=True)
print(age_list)

#Find Max age, Min age and sum of all ages from age_list.
print("Maximum Age:", max(age_list))
print("Minimum Age:", min(age_list))
print("Sum of All Ages:", sum(age_list))
```

```python
name_list = ["Aseeb", "Rahul", "Arun", "Vishnu", "Anand"]

#Print the first element
print("First Element:", name_list[0])

#Print the last element
print("Last Element:", name_list[-1])

#Print elements from index 2 to index 4
print("Elements from index 2 to 4:", name_list[2:5])

#Print elements in reverse order
print("Reversed List:", name_list[::-1])
```

```python
#Create a dictionary
student_marks = {
    "Aseeb": 85,
    "Afzal": 78,
    "Arun": 90,
    "Vishnu": 75,
    "Anand": 88
}

print("Student Marks:", student_marks)

#Access and print a specific student's mark
print("Afzal's Mark:", student_marks["Afzal"])

#Add a new student "Janani" with a mark of 80
student_marks["Janani"] = 80
print("After Adding Janani:", student_marks)

#Update the mark of any one older student to 82.
student_marks["Arun"] = 82
print("After Updating Arun's Mark:", student_marks)

#Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs.
print("Student Names:", student_marks.keys())
print("Student Marks:", student_marks.values())
print("Student Details:", student_marks.items())
```

```python
#Create a set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("My Set:", my_set)

#Attempt to modify an element
#my_set[4] = 's'

#Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)

#Union and Intersection
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

```python
#Attempt to modify an element
my_set[4] = 's'
Explanation:
- Sets do not support indexing because they are unordered collections.
- my_set[4] attempts to access an element using an index.
- Unlike lists, sets do not have fixed index positions.
- Therefore, Python raises a TypeError.
```

```python
# Performance Category Program

# 1. Get score from user
score = int(input("Enter your score (0-10): "))

# 2. Check whether the score is within the valid range
if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")

# 3. Determine performance category
elif score > 7:
    print("Performance Category: Above Average")
    print("Excellent performance! Keep up the good work.")

elif score >= 4:
    print("Performance Category: Average")
    print("Good effort! Keep practising to improve further.")

else:
    print("Performance Category: Below Average")
    print("Need to improve your performance. Consistent practice will lead to better results.")
```
